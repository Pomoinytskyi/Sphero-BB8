```python
from spherov2 import scanner
from spherov2.sphero_edu import SpheroEduAPI
from spherov2.toy.r2d2 import R2D2
import nest_asyncio
import asyncio

nest_asyncio.apply()

async def find_toy():
	toy =  scanner.find_toy()
	return toy

toy = asyncio.run(find_toy())
```

```python
import pygame
import time

pygame.init()
pygame.joystick.init()

# Check how many joysticks (controllers) are attached
joystick_count = pygame.joystick.get_count()
print(f"Number of joysticks: {joystick_count}")

if joystick_count == 0:
    print("No joystick connected. Exiting.")
    

# Select the first joystick
joystick = pygame.joystick.Joystick(0)
joystick.init()
print(f"Joystick name: {joystick.get_name()}")
```

```python
import math

try:
    with SpheroEduAPI(toy) as spheroApi:
        spheroApi.set_stabilization(True)
        spheroApi.reset_aim()
        while True:
            pygame.event.pump()

            x = joystick.get_axis(0)
            y = -1 * joystick.get_axis(1)

            if joystick.get_button(2):
                print("X button pressed. Exiting...")
                break

            if joystick.get_button(1):
                spheroApi.reset_aim()

            angle = math.degrees(math.atan2(x, y))
            angle = (angle + 360) % 360

            magnitude = math.sqrt(x*x + y*y)
            speed = int(255 * magnitude / 2)

            if magnitude < 0.1:
                speed = 0
                spheroApi.stop_roll()

            spheroApi.set_heading(int(angle))
            
            spheroApi.set_speed(speed)

            time.sleep(0.1)

except KeyboardInterrupt:
    print("Exiting control loop...")
finally:
    pygame.joystick.quit()
    pygame.quit()
```
